```python
#para criar uma função python

# def nome_da_função (parametros):
# sua função
#return resultado

# queremos criar uma funçao que é uma calculadora
def calculadora(num1, num2, operacao):
  if operacao == '+':
    return num1+num2
  elif operacao == '-':
    return num1-num2
  elif operacao == '*':
    return num1*num2
  elif operacao == '/': #não pode num dividido por zero
    if num2 == 0:
      return 'Erro, nao pode dividir por zero (0)'
    else:
      return num1 / num2
  else:
    return 'escolha um operador valido (+, -, * ou /)'
```

```python
# vamos simular a nossa calculadora
num1 = float(input('informe o primeiro numero\n'))
num2 = float(input('informe o segundo numero\n'))
operacao = input ( 'informe um operador\n')

calculadora(num1, num2, operacao)
```
