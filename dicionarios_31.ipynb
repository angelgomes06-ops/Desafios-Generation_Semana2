```python
dict_venda = {
    "id_transacao": 1001,
    "produto": "Smartphone",
    "categoria": "Eletronico",
    "preco": 2500.00,
    "quantidade": 2,
    "ativo": True
}

#tabela bão é num, é texto
# int = num string = texto float = r$ booleano = true/false
#podemos acessar os valores


print(dict_venda['produto'])
# usar apenas print(dict['chave] não é uma boa pratica)
print(dict_venda.get('desconto', "0%"))
# traz o resultado do dicionario, mas se nao encontra a funcao .get() por substrituir o erro, por algo informado

# outros métodos de ver dados
print(dict_venda.keys())
print(dict_venda.values())
print(dict_venda.items())

```

```python
# CRUD
# create
# read
# update
# delete

# vamos add uma nova chave - queremos a chave valor da compra
dict_venda['valor_compra'] = dict_venda['preco'] * dict_venda ['quantidade']
print(dict_venda['valor_compra'])

#queremos alterar um valor existente do dicionario
dict_venda['preco'] = 2600
print(dict_venda)

# .update() - atualizar ou adicionar muitos campos de uma vez
# dizer o que fazer
novos_dados = {
    "cidade": "São Paulo",
    "canal_venda": "online",
    "plataforma": "Shopee",
    "produto": "Smartphone Motorola"
    }
#fazer
dict_venda.update(novos_dados)
print(dict_venda)
```

```python
# vamos remover / deletar uma informação
canal = dict_venda.pop('canal_venda')
display(dict_venda)
```

```python
# vamos ver esses dados de maneira mais apresentaval (FOR)

for chave, valor in dict_venda.items():
  print(chave,valor)
```

```python
#dicionarios aninhados
dict_banco_clientes = {
    "clientes1": {
        "nome": "Luiz",
        "idade": 32,
        "compras": {"total_gasto":12000}
    },
    "clientes2":{
        "nome": "Felipe",
        "idade": 35,
        "compras": {"total_gasto":8000}
    }
}

# desafio simples
# acessar os dados do cliente 1
print(dict_banco_clientes['clientes1']['nome'])
# acessar o total de gasto
print(dict_banco_clientes["clientes1"]['compras']["total_gasto"])

# se temos um dicionario aninhado e queremos exbir valores iguais de
for chave, valor in dict_banco_clientes.items():
  nome = valor ['nome']
  print(nome)
```
